```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats
from datetime import datetime

print("Libraries imported successfully!")
```

```python
file_path = '../data/togo-dapaong_qc.csv'  
df = pd.read_csv(file_path)

print("Data loaded successfully!")
print(f"Dataset shape: {df.shape}")
```

```python
# Basic information about the dataset
print("=== FIRST 5 ROWS ===")
display(df.head())

print("\n=== DATASET INFO ===")
df.info()

print("\n=== SUMMARY STATISTICS ===")
display(df.describe())

print("\n=== MISSING VALUES ===")
missing_values = df.isna().sum()
print(missing_values)

# Identify columns with >5% nulls
total_rows = len(df)
cols_over_5perc_null = missing_values[missing_values / total_rows > 0.05]
print(f"\nColumns with >5% nulls: {list(cols_over_5perc_null.index)}")
```

```python
# Create a copy for cleaning
df_clean = df.copy()

print("Starting data cleaning...")

# 1. Handle missing values in KEY columns
key_columns = ['GHI', 'DNI', 'DHI', 'Tamb', 'RH', 'WS', 'BP']
for col in key_columns:
    if col in df_clean.columns and df_clean[col].isna().sum() > 0:
        median_val = df_clean[col].median()
        df_clean[col].fillna(median_val, inplace=True)
        print(f"Filled missing values in {col} with median: {median_val:.2f}")

# 2. Outlier Detection using Z-scores
print("\nDetecting outliers...")
numeric_cols = ['GHI', 'DNI', 'DHI', 'ModA', 'ModB', 'WS', 'WSgust']
# Only use columns that exist in the dataset
available_numeric_cols = [col for col in numeric_cols if col in df_clean.columns]

if available_numeric_cols:
    z_scores = np.abs(stats.zscore(df_clean[available_numeric_cols].dropna()))
    outliers = (z_scores > 3).any(axis=1)
    print(f"Number of outlier rows detected: {outliers.sum()}")
    
    # Remove outliers
    df_clean = df_clean[~outliers].copy()
    print(f"Shape after removing outliers: {df_clean.shape}")
else:
    print("No numeric columns available for outlier detection")

print("Data cleaning completed!")
```

```python
# Convert timestamp and extract time features
df_clean['Timestamp'] = pd.to_datetime(df_clean['Timestamp'])
df_clean['Hour'] = df_clean['Timestamp'].dt.hour
df_clean['Month'] = df_clean['Timestamp'].dt.month
df_clean['Day'] = df_clean['Timestamp'].dt.day

print("Timestamp conversion completed!")
print(f"Date range: {df_clean['Timestamp'].min()} to {df_clean['Timestamp'].max()}")
```

```python
# Plot 1: GHI Over Time
plt.figure(figsize=(15, 5))
plt.plot(df_clean['Timestamp'], df_clean['GHI'], alpha=0.7)
plt.title('GHI (Global Horizontal Irradiance) Over Time - Togo')
plt.xlabel('Timestamp')
plt.ylabel('GHI (W/m²)')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Plot 2: DNI Over Time
plt.figure(figsize=(15, 5))
plt.plot(df_clean['Timestamp'], df_clean['DNI'], alpha=0.7, color='orange')
plt.title('DNI (Direct Normal Irradiance) Over Time - Togo')
plt.xlabel('Timestamp')
plt.ylabel('DNI (W/m²)')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Plot 3: Temperature Over Time
plt.figure(figsize=(15, 5))
plt.plot(df_clean['Timestamp'], df_clean['Tamb'], alpha=0.7, color='red')
plt.title('Ambient Temperature Over Time - Togo')
plt.xlabel('Timestamp')
plt.ylabel('Temperature (°C)')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Analyze cleaning impact on solar modules
if 'Cleaning' in df_clean.columns:
    print("=== CLEANING IMPACT ANALYSIS ===")
    cleaning_impact = df_clean.groupby('Cleaning')[['ModA', 'ModB']].mean()
    print("Average module readings by cleaning status:")
    display(cleaning_impact)
    
    # Plot cleaning impact
    cleaning_impact.plot(kind='bar', figsize=(10, 6))
    plt.title('Impact of Cleaning on Module Readings - Togo')
    plt.xlabel('Cleaning Status (0=No, 1=Yes)')
    plt.ylabel('Average Reading (W/m²)')
    plt.legend(['Module A', 'Module B'])
    plt.show()
else:
    print("No 'Cleaning' column found in dataset")
```

```python
# Correlation Heatmap
corr_columns = ['GHI', 'DNI', 'DHI', 'TModA', 'TModB', 'Tamb', 'RH', 'WS', 'BP']
available_corr_cols = [col for col in corr_columns if col in df_clean.columns]

if len(available_corr_cols) > 1:
    corr_matrix = df_clean[available_corr_cols].corr()
    
    plt.figure(figsize=(12, 8))
    sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0, fmt='.2f')
    plt.title('Correlation Heatmap - Togo')
    plt.tight_layout()
    plt.show()
else:
    print("Not enough columns for correlation analysis")
```

```python
# Create subplots for multiple scatter plots
fig, axes = plt.subplots(2, 2, figsize=(15, 12))

# Plot 1: Wind Speed vs GHI
axes[0,0].scatter(df_clean['WS'], df_clean['GHI'], alpha=0.5)
axes[0,0].set_title('Wind Speed vs GHI')
axes[0,0].set_xlabel('Wind Speed (m/s)')
axes[0,0].set_ylabel('GHI (W/m²)')

# Plot 2: Relative Humidity vs GHI
axes[0,1].scatter(df_clean['RH'], df_clean['GHI'], alpha=0.5, color='green')
axes[0,1].set_title('Relative Humidity vs GHI')
axes[0,1].set_xlabel('Relative Humidity (%)')
axes[0,1].set_ylabel('GHI (W/m²)')

# Plot 3: Temperature vs GHI
axes[1,0].scatter(df_clean['Tamb'], df_clean['GHI'], alpha=0.5, color='red')
axes[1,0].set_title('Temperature vs GHI')
axes[1,0].set_xlabel('Temperature (°C)')
axes[1,0].set_ylabel('GHI (W/m²)')

# Plot 4: Barometric Pressure vs GHI
if 'BP' in df_clean.columns:
    axes[1,1].scatter(df_clean['BP'], df_clean['GHI'], alpha=0.5, color='purple')
    axes[1,1].set_title('Barometric Pressure vs GHI')
    axes[1,1].set_xlabel('Barometric Pressure (hPa)')
    axes[1,1].set_ylabel('GHI (W/m²)')

plt.tight_layout()
plt.show()
```

```python
# Histograms for key variables
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# GHI Distribution
axes[0,0].hist(df_clean['GHI'], bins=50, edgecolor='black', alpha=0.7)
axes[0,0].set_title('Distribution of GHI')
axes[0,0].set_xlabel('GHI (W/m²)')
axes[0,0].set_ylabel('Frequency')

# Wind Speed Distribution
axes[0,1].hist(df_clean['WS'], bins=50, edgecolor='black', alpha=0.7, color='orange')
axes[0,1].set_title('Distribution of Wind Speed')
axes[0,1].set_xlabel('Wind Speed (m/s)')
axes[0,1].set_ylabel('Frequency')

# Temperature Distribution
axes[1,0].hist(df_clean['Tamb'], bins=50, edgecolor='black', alpha=0.7, color='red')
axes[1,0].set_title('Distribution of Temperature')
axes[1,0].set_xlabel('Temperature (°C)')
axes[1,0].set_ylabel('Frequency')

# Relative Humidity Distribution
axes[1,1].hist(df_clean['RH'], bins=50, edgecolor='black', alpha=0.7, color='green')
axes[1,1].set_title('Distribution of Relative Humidity')
axes[1,1].set_xlabel('Relative Humidity (%)')
axes[1,1].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```
